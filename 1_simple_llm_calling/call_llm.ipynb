```python
import sounddevice as sd
import soundfile as sf
from IPython.display import Audio, display
from langchain_core.messages import SystemMessage, HumanMessage

def bolopay_voice():
    print("🎤 Speak in any supported language...")

    audio = sd.rec(
        5 * 16000,
        samplerate=16000,
        channels=1,
        dtype="float32",
        device=1
    )
    sd.wait()
    sf.write("bolopay_input.wav", audio, 16000)

    # Automatic language detection
    result = stt.transcribe(
        audio="bolopay_input.wav",
        language_code="auto"
    )

    transcript = result.get("transcript", "").strip()

    if not transcript:
        print("No speech recognized. Please try again.")
        return

    print("You:", transcript)

    response = llm.invoke([
        SystemMessage(content=f"""
You are BoloPay, a concise multilingual payment assistant.
Use only this knowledge:
{knowledge}

Answer in the same language as the user's question.
Keep the answer short.
Never ask for a UPI PIN or invent facts.
If unrelated or unsupported, reply with the appropriate translation of:
"I don't know. Is there anything I can help you with regarding your payments?"
"""),
        HumanMessage(content=transcript)
    ])

    answer = response.content
    print("BoloPay:", answer)

    # Use automatic TTS language selection
    tts.synthesize(
        answer,
        model="timbre-v2.5",
        language="auto",
        output_file="bolopay_response.wav"
    )

    display(Audio("bolopay_response.wav", autoplay=True))

bolopay_voice()
```

```python
from lingua import LanguageDetectorBuilder

detector = LanguageDetectorBuilder.from_all_languages().with_preloaded_language_models().build()

texts = [
    "meri bank ne kitani paise hai kitani paise hai mera bank mein",
    "मेरी बहन में कितनी पैसे है कितनी पैसे है मेरे बैंक में",
    "కితని పైసలు",
    "மேரி பேங்கில் கிட்னி பைசே மறை பேங்க்",
]

for text in texts:
    print(text, "→", detector.compute_language_confidence_values(text)[:3])
```

```python
candidates = {
    "en-IN": "meri bank ne kitani paise hai kitani paise hai mera bank mein",
    "hi-IN": "मेरी बहन में कितनी पैसे है कितनी पैसे है मेरे बैंक में",
    "te-IN": "కితని పైసలు",
    "ta-IN": "மேரி பேங்கில் கிட்னி பைசே மறை பேங்க்",
    "kn-IN": "ಬ್ಯಾಂಗ್ನ ಕಿತನಿ ಪೈಸೆ பைசே ಮರೆ ಬ್ಯಾಂಗ್",
    "ml-IN": "മേരി ബാങ്ക് കിട്ടി പൈസ ഇതിനെ പൈസ",
    "mr-IN": "मेरी बँगने कितनी पैसे आहे कितनी पैसे आहे मेरा बँक",
    "gu-IN": "મારી બેન્ડને કેટની પૈસે હૈ કેટલી પૈસા હૈ મારા બેન્ડમાં",
    "pa-IN": "ਮੇਰੀ ਬੈਗ ਕਿੰਨੀ ਪੈਸੇ ਹੈ ਕਿਤਨੀ ਪੈਸੇ ਹੈ ਮੇਰਾ ਬੈਗ"
}

best_text, best_lang, best_score = None, None, 0

for code, text in candidates.items():
    scores = detector.compute_language_confidence_values(text)
    top = scores[0]

    if top.value > best_score:
        best_text = text
        best_lang = code
        best_score = top.value

print("Language:", best_lang)
print("Confidence:", round(best_score * 100, 1), "%")
print("Transcript:", best_text)
```

```python
import os
import sounddevice as sd
import soundfile as sf
from IPython.display import Audio, display
from gnani.stt import GnaniSTTBatchClient
from gnani.tts import GnaniTTSClient
from langchain_core.messages import SystemMessage, HumanMessage

# Initialize Gnani clients
stt_batch = GnaniSTTBatchClient(api_key=os.getenv("GNANI_STT_API_KEY"))
tts = GnaniTTSClient(api_key=os.getenv("GNANI_TTS_API_KEY"))

# Supported language codes for TTS
tts_languages = {
    "en-IN": "en-IN",
    "hi-IN": "hi-IN",
    "te-IN": "te-IN",
    "ta-IN": "ta-IN",
    "kn-IN": "kn-IN",
    "ml-IN": "ml-IN",
    "bn-IN": "bn-IN",
    "mr-IN": "mr-IN",
    "gu-IN": "gu-IN",
    "pa-IN": "pa-IN",
}

def bolopay_voice():
    # 1. Record voice
    print("🎤 Speak now...")
    audio = sd.rec(
        int(5 * 16000),
        samplerate=16000,
        channels=1,
        dtype="float32",
        device=1
    )
    sd.wait()
    sf.write("bolopay_live.wav", audio, 16000)

    # 2. Detect language and transcribe
    print("🔎 Detecting language...")
    result = stt_batch.transcribe(
        files="bolopay_live.wav",
        language_code="auto"
    )

    if not result.files:
        print("No transcription result. Please try again.")
        return

    transcript_data = result.files[0].transcript
    if not isinstance(transcript_data, dict):
        print("Could not read transcription result:", transcript_data)
        return

    transcript = transcript_data.get("full_transcript", "").strip()
    language = transcript_data.get("language_code", "en-IN")

    if not transcript:
        print("No speech detected. Please try again.")
        return

    print(f"🌐 Detected language: {language}")
    print("You:", transcript)

    # 3. Ask the LLM
    response = llm.invoke([
        SystemMessage(content=f"""
You are BoloPay, a concise voice payment assistant.

Use only the following knowledge base to answer:
{knowledge}

Reply in the same language as the user's speech.
Keep your answer short and easy to understand when spoken aloud.
Never ask for a UPI PIN or invent facts.
Do not claim a payment succeeded unless the app confirms it.
If the request is unrelated to payments or unsupported by the knowledge base,
reply with the equivalent in the user's language of:
"I don't know. Is there anything I can help you with regarding your payments?"
If the request is relevant but unclear, ask a short clarification.
"""),
        HumanMessage(content=transcript)
    ])

    answer = response.content
    if not isinstance(answer, str):
        answer = str(answer)

    print("🤖 BoloPay:", answer)

    # 4. Speak the answer in the detected language
    if language not in tts_languages:
        print(f"TTS language {language} isn't configured in this example.")
        print("Please add a supported TTS language mapping.")
        return

    print("🔊 Speaking response...")
    tts.synthesize(
        answer,
        model="timbre-v2.5",
        language=tts_languages[language],
        output_file="bolopay_live_response.wav"
    )
    display(Audio("bolopay_live_response.wav", autoplay=True))

bolopay_voice()
```
